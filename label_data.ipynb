```python
# YouTube comments — TextBlob labeling
from pathlib import Path
import pandas as pd
from textblob import TextBlob

DATA_DIR = Path('data')
INPUT_PATH = DATA_DIR / 'youtube_comments.csv'
OUTPUT_PATH = DATA_DIR / 'youtube_comment_label.csv'
```

## Load cleaned data

```python
df = pd.read_csv(INPUT_PATH, usecols=['VideoTitle', 'CommentText'])
df['CommentText'] = df['CommentText'].fillna('').astype(str)
df = df[df['CommentText'].str.strip().ne('')].copy()
df.head()
```

## Label comments

```python
def classify_polarity(polarity: float) -> tuple[str, int]:
    if polarity > 0:
        return 'Positive', 2
    if polarity < 0:
        return 'Negative', 0
    return 'Neutral', 1

labeled = df[['VideoTitle', 'CommentText']].copy()
labeled['polarity'] = labeled['CommentText'].map(lambda text: float(TextBlob(text).sentiment.polarity))
labels = labeled['polarity'].map(classify_polarity)
labeled['Sentiment'] = labels.str[0]
labeled['Sentiment_label'] = labels.str[1].astype('int8')
labeled['Sentiment'].value_counts()
```

## Save labeled data

```python
labeled.to_csv(OUTPUT_PATH, index=False, encoding='utf-8-sig')
print(f'Saved: {OUTPUT_PATH}')
labeled.head()
```
